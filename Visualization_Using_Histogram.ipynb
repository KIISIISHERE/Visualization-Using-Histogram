```python
# Import all the necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files
```

```python
from google.colab import files

# Import the dataset and handle file upload
print("Please upload your Penguins dataset file:")
uploaded = files.upload()
file_name = list(uploaded.keys())
```

```python
import pandas as pd

# Load dataset into pandas
df = pd.read_csv(file_name[0])
```

```python
# Quick missing value cleanup to ensure charts render beautifully
numerical_cols = ['Culmen Length (mm)', 'Culmen Depth (mm)', 'Flipper Length (mm)', 'Body Mass (g)']
for col in numerical_cols:
    df[col] = df[col].fillna(df[col].median())

# Configure plot window style specifications
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_theme(style="whitegrid")

print("\n" + "="*50)
print("GENERATING CUSTOM BINNED HISTOGRAMS...")
print("="*50)

# Create a layout grid for the 4 quantitative features
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Penguin Biological Measurements Distribution Analysis', fontsize=16, fontweight='bold')
```

```python
import numpy as np
# 1 & 2. Plot Culmen Length (mm) with customized bins, range, and color
length_min, length_max = df['Culmen Length (mm)'].min(), df['Culmen Length (mm)'].max()
bins_length = np.arange(np.floor(length_min), np.ceil(length_max) + 2, 2) # 2mm step width
sns.histplot(ax=axes[0, 0], x='Culmen Length (mm)', data=df, bins=bins_length, color='royalblue', edgecolor='black', kde=True)
axes[0, 0].set_title('Distribution of Culmen Length', fontweight='bold')
axes[0, 0].set_xlabel('Culmen Length (mm)')
axes[0, 0].set_ylabel('Count')
```

```python
# 1 & 2. Plot Culmen Depth (mm) with customized bins, range, and color
depth_min, depth_max = df['Culmen Depth (mm)'].min(), df['Culmen Depth (mm)'].max()
bins_depth = np.arange(np.floor(depth_min), np.ceil(depth_max) + 0.5, 0.5) # 0.5mm step width
sns.histplot(ax=axes[0, 1], x='Culmen Depth (mm)', data=df, bins=bins_depth, color='crimson', edgecolor='black', kde=True)
axes[0, 1].set_title('Distribution of Culmen Depth', fontweight='bold')
axes[0, 1].set_xlabel('Culmen Depth (mm)')
axes[0, 1].set_ylabel('Count')
```

```python
# 1 & 2. Plot Flipper Length (mm) with customized bins, range, and color
flipper_min, flipper_max = df['Flipper Length (mm)'].min(), df['Flipper Length (mm)'].max()
bins_flipper = np.arange(np.floor(flipper_min), np.ceil(flipper_max) + 5, 5) # 5mm step width
sns.histplot(ax=axes[1, 0], x='Flipper Length (mm)', data=df, bins=bins_flipper, color='forestgreen', edgecolor='black', kde=True)
axes[1, 0].set_title('Distribution of Flipper Length', fontweight='bold')
axes[1, 0].set_xlabel('Flipper Length (mm)')
axes[1, 0].set_ylabel('Count')
```

```python
# 1 & 2. Plot Body Mass (g) with customized bins, range, and color
mass_min, mass_max = df['Body Mass (g)'].min(), df['Body Mass (g)'].max()
bins_mass = np.arange(np.floor(mass_min), np.ceil(mass_max) + 250, 250) # 250g step width
sns.histplot(ax=axes[1, 1], x='Body Mass (g)', data=df, bins=bins_mass, color='darkorange', edgecolor='black', kde=True)
axes[1, 1].set_title('Distribution of Body Mass', fontweight='bold')
axes[1, 1].set_xlabel('Body Mass (g)')
axes[1, 1].set_ylabel('Count')
plt.tight_layout()
plt.show()
```
